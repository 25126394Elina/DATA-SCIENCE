5. Create a random numpy array of size 6x3 and transpose it

```python
rand_arr=np.random.rand(6,3)

rand_arr_trans =rand_arr.T

print(rand_arr)
print(rand_arr_trans)
```

6. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 3, 4, 2

rand_arr1 = np.random.rand(n, p)
rand_arr2 = np.random.rand(p, m)

dot_product = np.dot(rand_arr1, rand_arr2)

print("Array 1 (n x p):")
print(rand_arr1)

print("\nArray 2 (p x m):")
print(rand_arr2)

print("\nDot Product:")
print(dot_product)
```
